```python
class Character():
    def __init__(self, username):
        self.__username = username
        self.__hp = 100
        self.__mana = 100
        self.__damage = 5
        self.__str = 0 # strength stat
        self.__vit = 0 # vitality stat
        self.__int = 0 # intelligence stat
        self.__agi = 0 # agility stat
    def getUsername(self):
        return self.__username
    def setUsername(self, new_username):
        self.__username = new_username
    def getHp(self):
        return self.__hp
    def setHp(self, new_hp):
        self.__hp = new_hp
    def getDamage(self):
        return self.__damage
    def setDamage(self, new_damage):
        self.__damage = new_damage
    def getStr(self):
        return self.__str
    def setStr(self, new_str):
        self.__str = new_str
    def getVit(self):
        return self.__vit
    def setVit(self, new_vit):
        self.__vit = new_vit
    def getInt(self):
        return self.__int
    def setInt(self, new_int):
        self.__int = new_int
    def getAgi(self):
        return self.__agi
    def setAgi(self, new_agi):
        self.__agi = new_agi
    def reduceHp(self, damage_amount):
        self.__hp = self.__hp - damage_amount
    def addHp(self, heal_amount):
        self.__hp = self.__hp + heal_amount

character1 = Character("Your Username")#Instance the object "Your Username" in the class Character
print(character1.getUsername())#prints the object
```

**The print(character1.getUsername()) is used to print the defined object in the class which is the class Character which the "Your Username" prints the output Your Username.**

#Single Inheritance

```python
class Novice(Character):
    def basicAttack(self, character):
        character.reduceHp(self.getDamage())
        print(f"{self.getUsername()} performed Basic Attack! -{self.getDamage()}")

character1 = Novice("Your Username") #Instance the object "Your Username" in the class Novice
print(character1.getUsername()) #prints the object
print(character1.getHp()) #prints the object's hp based on the class Character's Constructor
```

**The print(character1.getUsername()) is used to print the defined object in the class which is the class Novice which the "Your Username" prints the output Your Username and print(character.get hp()) is used to print the defined object's hp from the class Character's constructor which is 100.**

#Multi-Level Inheritance

**Swordsman.py**

```python
class Swordsman(Novice):
    def __init__(self, username):
        super().__init__(username)
        self.setStr(5)
        self.setVit(10)
        self.setHp(self.getHp()+self.getVit())

    def slashAttack(self, character):
        self.new_damage = self.getDamage()+self.getStr()
        character.reduceHp(self.new_damage)
        print(f"{self.getUsername()} performed Slash Attack! -{self.new_damage}")
```

**Archer.py**

```python
import random

class Archer(Novice):
    def __init__(self, username):
        super().__init__(username)
        self.setAgi(5)
        self.setInt(5)
        self.setVit(5)
        self.setHp(self.getHp()+self.getVit())

    def rangedAttack(self, character):
        self.new_damage = self.getDamage()+random.randint(0,self.getInt())
        character.reduceHp(self.new_damage)
        print(f"{self.getUsername()} performed Ranged Attack! -{self.new_damage}")
```

**Magician.py**

```python
class Magician(Novice):
    def __init__(self, username):
        super().__init__(username)
        self.setInt(10)
        self.setVit(5)
        self.setHp(self.getHp()+self.getVit())

    def heal(self):
        self.addHp(self.getInt())
        print(f"{self.getUsername()} performed Heal! +{self.getInt()}")

    def magicAttack(self, character):
        self.new_damage = self.getDamage()+self.getInt()
        character.reduceHp(self.new_damage)
        print(f"{self.getUsername()} performed Magic Attack! -{self.new_damage}")
```

**Boss.py**

```python
class Boss(Swordsman, Archer, Magician): # multiple inheritance
    def __init__(self, username):
        super().__init__(username)
        self.setStr(10)
        self.setVit(25)
        self.setInt(5)
        self.setHp(self.getHp()+self.getVit())
```

#SUPPLEMENTARY ACTIVITY


**Requirements:**
1. The game must be able to select between 2 modes: Single player and Player vs Player. The game can spawn
multiple matches where single player or player vs player can take place.
2. In Single player:
• the player must start as a Novice, then after 2 wins, the player should be able to select a new role between
Swordsman, Archer, and Magician.
• The opponent will always be a boss named Monster.
3. In Player vs Player, both players must be able to select among all the possible roles available except Boss.
4. Turns of each player for both modes should be randomized and the match should end when one of the players hp is
zero.
5. Wins of each player in a game for both the modes should be counted

```python
# randomness
import random


# Main Menu User Input
print("\n ---MAIN MENU---")


def main():
    while True:
        # Display menu options
        print("Select Game Mode: ")
        print("1. Single Player")
        print("2. PVP")
        print("3. Exit")

        # Prompt user for game mode selection
        choice = input("Enter Choice (1/2/3): ")

        # Route to selected mode based on user input
        if choice == "1":
            single_plyr_mode()
        elif choice == "2":
            pvp_mode()
        elif choice == "3":
            print("Exiting the game...")
            break  # Exit the main menu loop
        else:
            print("Invalid Choice")


# SinglePlayerMode Function
def single_plyr_mode():
    player_1 = input("Enter Player 1 Name: ")

    # Initialize player as a basic Novice role
    player = Novice(player_1)

    # Track player statistics
    wins = 0
    matches_played = 0

    while True:
        # Create a new boss opponent for every match
        opponent = Boss("Destructor")

        # Simulate the battle
        result = play_a_match(player, opponent)
        matches_played += 1

        # Check if the player won the match
        if result == player:
            wins += 1
            print(f"You won the match! Total Wins: {wins}")
            matches_played += 1

            # Role upgrade after 2 wins
            if wins == 2:
                print(
                    "Choose a new role: 1. Swordsman, 2. Archer, 3. Magician"
                )
                role_choice = input("Enter Choice (1/2/3): ")

                # Instantiate new role while maintaining the existing username
                if role_choice == "1":
                    player = Swordsman(player.getUsername())
                elif role_choice == "2":
                    player = Archer(player.getUsername())
                elif role_choice == "3":
                    player = Magician(player.getUsername())
                else:
                    print("Invalid Choice stays as the current role.")

        # Stop automatically after 3 matches
        if matches_played >= 3:
            print("Game Over!")
            print(f"Total Wins: {wins}")
            break

        # Ask if player wants another match
        if input("Play another match? (y/n): ").lower() != "y":
            break


# Player vs Player Mode Function
def pvp_mode():

    # Player 1 setup: name and role selection
    player1 = input("Enter Player 1 Name: ")
    print(f"{player1}, choose your role: 1. Swordsman, 2. Archer, 3. Magician")
    role1 = input("Enter your role (1/2/3): ")
    player1 = create_role(role1, player1)

    # Player 2 setup: name and role selection
    player2 = input("Enter Player 2 Name: ")
    print(f"{player2}, choose your role: 1. Swordsman, 2. Archer, 3. Magician")
    role2 = input("Enter your role (1/2/3): ")
    player2 = create_role(role2, player2)

    # Track overall PvP score
    wins_p1 = 0
    wins_p2 = 0
    play_again = "y"

    # PvP game loop
    while play_again == "y":
        # Display starting HP for both players
        print(f"{player1.getUsername()} HP: {player1.getHp()}")
        print(f"{player2.getUsername()} HP: {player2.getHp()}")

        # Run the match
        winner = play_a_match(player1, player2)

        # Update win counters based on winner
        if winner == player1:
            wins_p1 += 1
            print(
                f"{player1.getUsername()} wins this round! Total Wins:"
                f" {wins_p1}"
            )
        else:
            wins_p2 += 1
            print(
                f"{player2.getUsername()} wins this round! Total Wins:"
                f" {wins_p2}"
            )

        # Prompt for another match
        play_again = input("Play another match? (y/n): ").lower()

    # loop ends automatically when play_again != "y"
    # Display overall match summary and declare the series champion
    print("\n=== Final PvP Score ===")
    print(f"{player1.getUsername()} Wins: {wins_p1}")
    print(f"{player2.getUsername()} Wins: {wins_p2}")

    if wins_p1 > wins_p2:
        print(f"Overall Winner: {player1.getUsername()}")
    elif wins_p2 > wins_p1:
        print(f"Overall Winner: {player2.getUsername()}")
    else:
        print("It's a tie!")


# Create a Role Function
def create_role(role, name):
    if role == "1":
        return Swordsman(name)
    elif role == "2":
        return Archer(name)
    elif role == "3":
        return Magician(name)
    else:
        # Default fallback role if choice is invalid
        return Novice(name)


# Simulate a Match Function
def play_a_match(p1, p2):
    players = [p1, p2]
    round_num = 1

    print(f"\n Match Start: {p1.getUsername()} vs {p2.getUsername()}")
    print("====================================")

    # Fight loop continues as long as both fighters are alive
    while p1.getHp() > 0 and p2.getHp() > 0:
        print(f"\n--- Round {round_num} ---")

        # Randomize turn order for fairness
        random.shuffle(players)
        attacker, defender = players

        # Perform class-specific action for the attacker
        if isinstance(attacker, Swordsman):
            attacker.slashAttack(defender)
        elif isinstance(attacker, Archer):
            attacker.rangedAttack(defender)
        elif isinstance(attacker, Magician):
            # Magician can heal or attack randomly in the game
            if random.choice(["attack", "heal"]) == "heal":
                attacker.heal()
            else:
                attacker.magicAttack(defender)
        else:
            attacker.basicAttack(defender)

        # Output current status after turn execution
        print(
            f"{attacker.getUsername()} attacks {defender.getUsername()}! "
            f"{defender.getUsername()} HP left: {defender.getHp()}"
        )
        print(
            f"{defender.getUsername()} attacks {attacker.getUsername()}! "
            f"{attacker.getUsername()} HP left: {attacker.getHp()}"
        )
        print("====================================")
        round_num += 1

    # Determine match winner
    if p1.getHp() > 0:
        winner = p1
    else:
        winner = p2

    # Print summary of match result
    print("\n=== Match Summary ===")
    print(f"{p1.getUsername()} HP: {p1.getHp()}")
    print(f"{p2.getUsername()} HP: {p2.getHp()}")
    print(f"Winner: {winner.getUsername()}")
    print("=====================")

    return winner


# Execution Code for the Main Driver code
if __name__ == "__main__":
    # Start the game loop
    main()
```
