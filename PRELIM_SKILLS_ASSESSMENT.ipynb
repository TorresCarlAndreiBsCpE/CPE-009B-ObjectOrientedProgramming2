```python

#Parent Class (For Parsing and Reading txt file)
class RecipeCookbook:
    #Constructor
    def __init__(self, filename="recipes.txt"):
        #Store file reference and immediately populate recipes list by instantiation
        self.filename = filename
        self.recipes = self.load_recipes()

    #Method called by __init__ to read and parse recipes from the text file
    def load_recipes(self):
        recipes = []
        try:
            #utf-8 prevents special character issues
            with open(self.filename, "r", encoding="utf-8") as f:
                #Split the content into distinct recipe blocks separated by double newlines
                content = f.read().strip().split("\n\n")

            for block in content:
                if not block.strip():
                    continue
                r = {}
                instructions_lines = []
                is_reading_instructions = False

                #Parse each block line-by-line using key prefixes
                for line in block.strip().split("\n"):
                    if line.startswith("ID: "):
                        r["id"] = int(line.split(": ", 1)[1])
                    elif line.startswith("Title: "):
                        r["title"] = line.split(": ", 1)[1]
                    elif line.startswith("Category: "):
                        r["category"] = line.split(": ", 1)[1]
                    elif line.startswith("Ingredients: "):
                        r["ingredients"] = line.split(": ", 1)[1]
                    elif line.startswith("Instructions: "):
                       text = line.split(": ", 1)[1].strip()
                       if text:
                           instructions_lines.append(text)
                       #Set a boolean flag to capture subsequent lines as multiline instructions
                       is_reading_instructions = True
                    elif is_reading_instructions:
                      if line.strip():
                       instructions_lines.append(line.strip())

                #Format accumulated multiline instructions with line breaks
                if instructions_lines:
                  r["instructions"] = "\n".join(instructions_lines)

                if r:
                    recipes.append(r)
        except FileNotFoundError:
            #print Error if the file does not exist on the initial commit
            print(f"Error: '{self.filename}' not found.")
        return recipes

#Child Class for UI Layer
class RecipeOrganizer(RecipeCookbook):
    #Constructor
    def __init__(self, filename="recipes.txt"):
        #Explicit call to Parent Constructor via super()
        super().__init__(filename)

    #Main Menu Display Output
    def display_menu(self):
        print("\n" + "=" * 40)
        print("        RECIPE ORGANIZER")
        print("=" * 40)
        print("1. View All Recipes")
        print("2. View Specific Recipe Details")
        print("3. Add a New Recipe")
        print("4. Exit")
        print("-" * 40)

    #View all of the recipes
    def view_all_recipes(self):
        print("\n--- ALL RECIPES ---")
        if not self.recipes:
            print("No recipes available.")
            return
        #Proper Formatting of the Rows and Columns using left-aligned string formatters
        print(f"{'ID':<4} | {'Title':<25} | {'Category':<12}")
        print("-" * 46)
        for r in self.recipes:
            print(
                f"{r.get('id', ''):<4} | {r.get('title', ''):<25} | {r.get('category', ''):<12}"
            )
    #view the recipe detail
    def view_recipe_detail(self):
        self.view_all_recipes() #Calls view_all_recipes to print the list before prompting for input
        if not self.recipes:
            return
        #User Input for the ID of the Recipe
        try:
            recipe_id = int(
                input("\nEnter the ID of the recipe you want to view: ")
            )
            #search for the recipe dict by its ID using a generator expression
            recipe = next(
                (r for r in self.recipes if r.get("id") == recipe_id), None
            )

            if recipe:
                #formatting the outputs of the recipes
                print("\n" + "*" * 40)
                print(f"RECIPE #{recipe['id']}: {recipe['title'].upper()}")
                print("*" * 40)
                print(f"Category   : {recipe['category']}")
                print(f"Ingredients: {recipe['ingredients']}")
                print(f"Instructions:\n{recipe['instructions']}")
                print("*" * 40)
            else:
                print("Error: Recipe ID not found.")
        except ValueError:
          # Handle non-integer user input attempts safely
            print("Error: Please enter a valid numerical ID.")

    #function to add a new recipe
    def add_recipe(self):
        print("\n--- ADD A NEW RECIPE ---")
        title = input("Enter recipe title: ").strip()
        if not title:
            print("Error: Title cannot be empty.")
            return

        # Restricted Category Selection for Food Categories and Drinks too
        print("\n==================")
        print("\nSelect Category:")
        print("1. Main Dish")
        print("2. Appetizer")
        print("3. Dessert")
        print("4. Soup")
        print("5. Noodles")
        print("6. Pasta")
        print("7. Drink")
        cat_choice = input("Enter choice (1-7): ").strip()

        if cat_choice == "1":
            category = "Main Dish"
        elif cat_choice == "2":
            category = "Appetizer"
        elif cat_choice == "3":
            category = "Dessert"
        elif cat_choice == "4":
            category = "Soup"
        elif cat_choice == "5":
            category = "Noodles"
        elif cat_choice == "6":
            category = "Pasta"
        elif cat_choice == "7":
            category = "Drink"
        else:
            print("Error: Invalid choice. Defaulting to 'Main Dish'.")
            category = "Main Dish"

        #Provides a loop input for the ingredients until the user input is Done
        print("Enter the Ingredients of the Dish (Type Done when it's complete):")
        ingredient_list = []
        while True:
            ingredient = input().strip()
            if ingredient.lower() == "done":
                break
            if ingredient:
                ingredient_list.append(ingredient)

        ingredients = ", ".join(ingredient_list) if ingredient_list else "No Ingredients Provided"

        #Provides a loop input for the instructions with Dynamic Numbering until the user input is Done
        print("Enter the Instructions of the Dish (Type Done when it's complete):")
        instruction_list = []
        step_number = 1
        while True:
            step = input(f"Step {step_number}: ").strip()
            if step.lower() == "done":
                break
            if step:
                instruction_list.append(f"{step_number}. {step.strip()}")
                step_number += 1

        instructions = "\n".join(instruction_list) if instruction_list else "No Instructions Provided."

        #provide a new data to allocate the given info
        new_id = max([r.get("id", 0) for r in self.recipes], default=0) + 1
        new_recipe = {
            "id": new_id,
            "title": title,
            "category": category,
            "ingredients": ingredients,
            "instructions": instructions,
        }

        # Update the runtime list in the memory
        self.recipes.append(new_recipe)

        #write a new custom recipe by the user input by append mode 'a'
        with open(self.filename, "a", encoding="utf-8") as f:
            f.write(f"ID: {new_recipe['id']}\n")
            f.write(f"Title: {new_recipe['title']}\n")
            f.write(f"Category: {new_recipe['category']}\n")
            f.write(f"Ingredients: {new_recipe['ingredients']}\n")
            f.write(f"Instructions: {new_recipe['instructions']}\n\n")

        print(f"Success! '{title}' has been added.")

    #Main Menu Display Driver Code
    def run(self):
        while True:
            self.display_menu()
            choice = input("Select an option (1-4): ").strip()

            if choice == "1":
                self.view_all_recipes()
            elif choice == "2":
                self.view_recipe_detail()
            elif choice == "3":
                self.add_recipe()
            elif choice == "4":
                print("\nThank you for using Recipe Organizer. Goodbye!")
                break
            else:
                print("Invalid selection. Please enter 1, 2, 3, or 4.")

#Script Entry Point Execution
if __name__ == "__main__":
    #instantiate child class
    app = RecipeOrganizer()
    app.run()
```
